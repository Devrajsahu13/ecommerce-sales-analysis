   # Loading Olist data into MySQL
   Reads the 8 Olist CSV files, converts the date columns, and loads each one into a MySQL table.

```python
import pandas as pd
from sqlalchemy import create_engine

engine = create_engine("mysql+pymysql://root:YOUR_PASSWORD@localhost:3306/olist")

files = {
    "customers": "olist_customers_dataset.csv",
    "orders": "olist_orders_dataset.csv",
    "order_items": "olist_order_items_dataset.csv",
    "payments": "olist_order_payments_dataset.csv",
    "reviews": "olist_order_reviews_dataset.csv",
    "products": "olist_products_dataset.csv",
    "sellers": "olist_sellers_dataset.csv",
    "category_translation": "product_category_name_translation.csv",
}

for table, file in files.items():
    df = pd.read_csv(f"data/{file}", encoding="utf-8-sig")
    for col in df.columns:
        if "timestamp" in col or col.endswith("_date") or col.endswith("_at"):
            df[col] = pd.to_datetime(df[col], errors="coerce")
    df.to_sql(table, engine, if_exists="replace", index=False, chunksize=5000)
    print(f"{table}: {len(df):,} rows")
```
